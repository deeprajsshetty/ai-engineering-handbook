## Text to Speech (TTS) API

```python
# Load the OpenAI API key from a .env file in the project directory.
# The .env file should contain a single line: OPENAI_API_KEY="your-openai-api-key"   (dummy placeholder; put your real key in handbook/.env (template: handbook/.env.example))
from dotenv import load_dotenv, find_dotenv

load_dotenv(find_dotenv(), override=True)
```

```python
# text = 'Viitorul este o carte deschisă, plină de capitole încă nescrise,\
# așteptând strălucirea minților noastre să picteze poveștile lor.'

text = 'In the future, we may be able to travel to other galaxies using wormholes or warp drives.'
```

```python
cost = len(text)/1000 * 0.15  # cost in cents
cost
```

```python
from openai import OpenAI
client = OpenAI()

# 1. Recommended: use with_streaming_response to properly stream audio to a file.
with client.audio.speech.with_streaming_response.create(
    model="tts-1",
    voice="alloy",
    input=text
) as response:
    response.stream_to_file('tts.mp3')

# 2. Alternative: create the response object first, then write the audio bytes at once.
response = client.audio.speech.create(
    model="tts-1",
    voice="alloy",
    input=text
)

# Deprecated: triggers DeprecationWarning — doesn't actually stream the content.
# response.stream_to_file('tts.mp3')

with open('tts-1.mp3', 'wb') as f:
    f.write(response.content)
```
