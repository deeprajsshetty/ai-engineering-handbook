## Transcriptions with Whisper

```python
# Load the OpenAI API key from a .env file in the project directory.
# The .env file should contain a single line: OPENAI_API_KEY="your-openai-api-key"   (dummy placeholder; put your real key in handbook/.env (template: handbook/.env.example))
from dotenv import load_dotenv, find_dotenv

load_dotenv(find_dotenv(), override=True)
```

```python
from openai import OpenAI
client = OpenAI()

file = 'rr.mp3'
with open(file, 'rb') as audio_file:
    transcript = client.audio.transcriptions.create(
        model='whisper-1',
        file=audio_file
    )
    print(transcript)
```

```python
print(transcript.text)
```

## Translations with Whisper

```python

from openai import OpenAI
client = OpenAI()

with open('german.mp3', 'rb') as audio_file:
    translation = client.audio.translations.create(
        model='whisper-1',
        file=audio_file
    )

print(translation)
```

```python
print(translation.text)
```
