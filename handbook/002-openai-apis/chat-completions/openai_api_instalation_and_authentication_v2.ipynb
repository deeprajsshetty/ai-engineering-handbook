## OpenAI API Authentication

You must acquire an OpenAI API Key and set it as an environment variable, prompt the user for it, or load it from a file.

1. Sign up at: https://platform.openai.com/signup
2. Generate a new API Key


#### Use ONE of the following ways to authenticate to OpenAI


```python
import os
# 1. Use enviroment variables in .dotenv (the recommended way)
# a) create a file called .env in the current directory
# b) add to the file: OPENAI_API_KEY="your-openai-api-key"   # dummy placeholder; put your real key in handbook/.env (template: handbook/.env.example)

# loading the OpenAI API key
from dotenv import load_dotenv, find_dotenv
load_dotenv(find_dotenv(), override=True)
```

```python
from openai import OpenAI
api_key = os.getenv('OPENAI_API_KEY')
client = OpenAI()
```

```python
# Prompts the user to paste their key securely (input is hidden, like a password).
# The assertion checks it starts with 'sk-' to catch copy-paste mistakes early.
# 2. Use getpass and prompt the user for the API Key 
import getpass

api_key = getpass.getpass('Paste your API Key: ')
assert api_key.startswith('sk-'), 'Error loding the API key. OpenAI API Keys start with "sk-".'
client = OpenAI(api_key=api_key)
```

```python
# Reads the API key from a plain-text file.
# The file should contain only the key on the first line.
# 3. Load the key from a file
api_key = open('key.txt', 'r').read().strip('\n')
assert api_key.startswith('sk-'), 'Error loding the API key. OpenAI API Keys start with "sk-".'
client = OpenAI(api_key=api_key)
```

## Testing the Authentication

```python
# Tests that authentication worked by listing available models.
# A successful response means your API key is valid and active.
# display all available models
for model in client.models.list():
    print(model)
```
