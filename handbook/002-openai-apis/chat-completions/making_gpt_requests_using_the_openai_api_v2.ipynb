```python
import os

# NOTE: Hardcoding API keys is not recommended — your key can be accidentally
# shared or committed to version control. The recommended approach is to store
# the key in a .env file (OPENAI_API_KEY=your-key-here) and load it with:
#   from dotenv import load_dotenv
#   load_dotenv()
os.environ['OPENAI_API_KEY'] = os.getenv('OPENAI_API_KEY', '')  # set it in .env, never hard-code keys
# Verify the key was loaded into the environment:
os.getenv('OPENAI_API_KEY')
```

```python
# Lists all models available to your API account.
# Useful to verify your key works and to discover model names.
from openai import OpenAI
client = OpenAI()

# display all available models
for model in client.models.list():
    print(model)
```

```python
# Send a request that generates TWO independent completions (n=2).
# Each choice in response.choices is an independent reply from the model.
response = client.chat.completions.create(
    model='gpt-5-mini',
    messages=[
        {'role': 'system', 'content': 'You are a helpful assistant.'},
        {'role': 'user', 'content': 'Write a random joke.'}
    ],
    n=2
)
# roles => system, user, assistant
print(response)
```

```python
# Print the text of the first (index 0) generated reply.
print(response.choices[0].message.content)
```

```python
# Print the second (index 1) reply — only available because n=2 was set.
print(response.choices[1].message.content)
```
