## Building Chat Conversations


```python
# Import required libraries for API access and environment variable loading.
from google import genai
import os
from dotenv import load_dotenv, find_dotenv

# Load environment variables from the .env file to retrieve the API key.
load_dotenv(find_dotenv(), override=True)
os.environ.get('GOOGLE_API_KEY')

# Initialize the Gemini client with the API key.
client = genai.Client(api_key=os.environ.get('GOOGLE_API_KEY'))
```

```python
# Create a ChatSession — the SDK automatically tracks and sends the full
# conversation history on each turn, enabling multi-turn conversations.
chat = client.chats.create(model='gemini-flash-latest')
chat
```

```python
# Send the first message to start the conversation.
response = chat.send_message('In one paragraph, explain how the Internet works to a beginner.')
```

```python
# Display the full conversation history as a list of Content objects.
chat.get_history()
```

```python
# Display the first entry in the history — the user's opening message.
chat.get_history()[0]
```

```python
# Display the second entry in the history — the model's reply.
chat.get_history()[1]
```

```python
# Extract the plain text from the model's first response.
chat.get_history()[1].parts[0].text
```

```python
# Continue the conversation — the model retains the full context of previous messages.
response = chat.send_message('Okay, how about a more detailed explanation about the IP addresses?')
```

```python
# Inspect the updated history — it now contains four entries (two full exchanges).
chat.get_history()
```

```python
# Ask a meta-question — the model can summarise what it has been asked so far.
response = chat.send_message('What have I asked you so far')
chat.get_history()
```

```python
# Print only the most recent model response from the history.
print(chat.get_history()[-1].parts[0].text)
```

```python
# Print the full conversation in a readable format by iterating over the history.
for item in chat.get_history():
    print(f'{item.role.capitalize()}: {item.parts[0].text}')
    print('-' * 100)
```

```python
# Import required libraries for API access and environment variable loading.
from google import genai
import os
from dotenv import load_dotenv, find_dotenv

# Load environment variables from the .env file to retrieve the API key.
load_dotenv(find_dotenv(), override=True)
os.environ.get('GOOGLE_API_KEY')

# Initialize the Gemini client with the API key.
client = genai.Client(api_key=os.environ.get('GOOGLE_API_KEY'))
```

```python
import time

# Create a new chat session for the interactive loop.
chat = client.chats.create(model='gemini-flash-latest')

# Run an interactive chat loop — type 'exit', 'quit', or 'bye' to end the session.
while True:
    prompt = input('User: ')
    if prompt.lower() not in ['exit', 'quit', 'bye']:
        response = chat.send_message(prompt)
        history = chat.get_history()
        print(f'{history[-1].role.capitalize()}: {history[-1].parts[0].text}')
        print('\n' + '*' * 100 + '\n')
    else:
        print('Quitting ...')
        time.sleep(2)
        print('Bye-bye!')
        break
```
