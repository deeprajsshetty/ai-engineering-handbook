## Gemini API Generation Parameters

```python
# Import required libraries for API access and environment variable loading.
from google import genai
from google.genai import types
import os
from dotenv import load_dotenv, find_dotenv

# Load environment variables from the .env file to retrieve the API key.
load_dotenv(find_dotenv(), override=True)

# Create the Gemini client with the API key.
client = genai.Client(api_key=os.environ.get('GOOGLE_API_KEY'))

MODEL = 'gemini-flash-latest'

# Define the prompt and generate a response from the model.
prompt = 'Rewrite "The Raven" by Edgar Allan Poe in the style of a modern rock song.'
response = client.models.generate_content(model=MODEL, contents=prompt)

# Print the generated text response.
print(response.text)
```

```python
# Create a GenerateContentConfig object to control the model's output behavior.
generation_config = types.GenerateContentConfig()  # Uses all default values.

# Example: Create a config with custom parameter values (uncomment to use).
# generation_config = types.GenerateContentConfig(
#     stop_sequences=[','],
#     max_output_tokens=32000,
#     temperature=0.9,
#     top_p=1,
#     top_k=1
# )

# Pass the config via the 'config' parameter of generate_content().
response = client.models.generate_content(
    model=MODEL,
    contents=prompt,
    config=generation_config
)
```

## Gemini API Generation  Parameters Explained

```python
generation_config = types.GenerateContentConfig(
    # candidate_count=1,  # Most models only support candidate_count=1
    # stop_sequences=[';'],
    # max_output_tokens=32000,
    temperature=0.0,
    top_k=4,
    top_p=1
    # Example: top_p=0.5 with token probs A=0.3, B=0.2, C=0.1 → selects A and B.
)

# Temperature=0 makes the model deterministic — it always picks the most likely token.
generation_config_temp_0 = types.GenerateContentConfig(temperature=0.0)

# Temperature=1 makes the model more creative and varied in its responses.
generation_config_temp_1 = types.GenerateContentConfig(temperature=1.0)

prompt = 'What is the meaning of life according to Viktor Frankl?'

# Generate a response with temperature=0 (deterministic, consistent output).
response_temp_0 = client.models.generate_content(
    model=MODEL, contents=prompt, config=generation_config_temp_0
)

# Generate a response with temperature=1 (creative, more varied output).
response_temp_1 = client.models.generate_content(
    model=MODEL, contents=prompt, config=generation_config_temp_1
)

print(response_temp_0.text)
print('\n' + '*' * 100 + '\n')
print(response_temp_1.text)
```
