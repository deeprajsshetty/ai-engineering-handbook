## Aggregations and Grouping Data

```python
import pandas as pd
# Sample DataFrame
data = {
    'Category': ['Electronics', 'Electronics', 'Furniture', 'Furniture', 'Apparel'],
    'Product': ['Laptop', 'Mouse', 'Chair', 'Table', 'Shoes'],
    'Sales': [1200, 100, 400, 300, 50]
}

df = pd.DataFrame(data)
```

```python
df
```

```python
grouped = df.groupby('Category')['Sales'].sum().reset_index()
grouped
```

```python
grouped = df.groupby(['Category', 'Product'])['Sales'].sum()
grouped
```

```python
summary = df.groupby(['Category', 'Product'])['Sales'].agg(['sum', 'mean', 'count'])
summary
```

## Project: Analyzing Website Traffic Data

```python
import pandas as pd
import numpy as np

# Set random seed for reproducibility
np.random.seed(42)

# Number of records
num_records = 1000

# Traffic sources
sources = ['Organic', 'Direct', 'Paid', 'Social', 'Referral']

# Traffic source distribution
source_probabilities = [0.40, 0.20, 0.15, 0.15, 0.10]

# Generate traffic data
data = {
    'Source': np.random.choice(
        sources,
        size=num_records,
        p=source_probabilities
    ),
    'SessionDuration': np.round(
        np.random.uniform(1.0, 30.0, num_records),
        2
    )
}

# Create Pandas DataFrame
df = pd.DataFrame(data)

# Save DataFrame to CSV
df.to_csv('traffic_data.csv', index=False)

print("Traffic data generated successfully!")
print(f"Total Records: {len(df)}")

# Display first 20 records
df.head(20)
```

```python
import pandas as pd
import numpy as np

# Set random seed for reproducibility
np.random.seed(42)

# Number of records
num_records = 1000

# Define traffic sources
sources = ['Direct', 'Referral', 'Social', 'Paid', 'Organic']

# Define traffic source probabilities
source_probabilities = [0.20, 0.20, 0.20, 0.15, 0.25]

# Generate traffic data
data = {
    'Source': np.random.choice(
        sources,
        size=num_records,
        p=source_probabilities
    ),

    'SessionDuration': np.round(
        np.random.uniform(1.0, 30.0, num_records),
        2
    ),

    'Visits': np.random.randint(
        1,
        500,
        size=num_records
    ),

    'Date': pd.date_range(
        start='2023-01-01',
        periods=num_records,
        freq='D'
    )
}

# Create DataFrame
df = pd.DataFrame(data)

# Save DataFrame to CSV
df.to_csv('website_traffic_data_datetime.csv', index=False)

print("Traffic dataset generated successfully!")
print("Total Records:", len(df))
print("Total Columns:", len(df.columns))

# Display first 20 records
df.head(20)
```
